## Les fonctions de perte

Lorsqu'on entraîne un réseau de neurones, l'objectif est de minimiser l'erreur entre les prédictions du modèle et les valeurs réelles. Ce processus repose sur trois composants clés : les fonctions de perte, les optimiseurs et la descente de gradient.

#### Qu'est-ce qu'une fonction de perte ?

Une fonction de perte (ou fonction de coût) mesure à quel point les prédictions d'un réseau de neurones correspondent aux valeurs réelles. Elle quantifie l'erreur entre la sortie prédite et la sortie réelle : elle indique au réseau à quel point il se trompe.

L'objectif de l'entraînement est de minimiser cette erreur — la rendre la plus petite possible.

#### Fonctions de perte courantes

- Mean Squared Error (MSE) : utilisée pour les tâches de régression (ex : prédire le prix d'une maison). Mesure la moyenne des écarts au carré entre prédictions et valeurs réelles.
- Categorical Crossentropy : utilisée pour les tâches de classification multiclasse (ex : reconnaître des chiffres manuscrits). Mesure à quel point les probabilités prédites correspondent aux labels réels.
- Binary Crossentropy : utilisée pour les tâches de classification binaire (ex : détecter une transaction frauduleuse). Similaire à la crossentropie catégorielle, mais pour deux classes.

````python
MSE = (1/n) * somme((y_reel - y_predit) ** 2)
````

Le réseau prédit qu'une maison coûtera 300 000 $. Le prix réel est 350 000 $. La fonction de perte calcule l'écart, par exemple avec le MSE :

Plus l'erreur est grande, plus la perte est élevée.

## Les optimiseurs : corriger les erreurs

Un optimiseur est un algorithme qui aide un modèle de machine learning à ajuster ses paramètres (poids et biais) pour minimiser la perte et améliorer sa performance. C'est lui qui détermine comment le modèle met à jour ses variables internes en fonction des erreurs qu'il commet.

Sans optimiseur, le modèle ne saurait pas comment ajuster ses poids pour s'améliorer.

#### Comment fonctionnent les optimiseurs ?

- Calculer la perte : le modèle fait des prédictions, et l'écart avec les valeurs réelles est calculé.
- Calculer les gradients (rétropropagation) : l'optimiseur calcule la contribution de chaque poids à la perte, via les dérivées partielles.
- Mettre à jour les poids : l'optimiseur ajuste les poids selon une règle d'apprentissage (comme la descente de gradient). Cette étape se répète sur de nombreuses itérations (epochs) jusqu'à convergence.

#### Types d'optimiseurs

- Descente de gradient (Gradient Descent) : met à jour les poids en se déplaçant dans la direction de la pente la plus forte de la fonction de perte. Simple mais lent sur de gros datasets.
- Descente de gradient stochastique (SGD) : met à jour les poids après chaque exemple d'entraînement plutôt que sur tout le dataset. Plus rapide, mais plus bruité.
- Mini-Batch Gradient Descent : équilibre entre full-batch et SGD, en mettant à jour les poids par petits lots.
- Optimiseurs à momentum (Momentum, NAG) : utilisent une moyenne mobile des gradients passés pour accélérer l'apprentissage et éviter les minima locaux.
- Optimiseurs adaptatifs (AdaGrad, RMSprop, Adam, AdamW) : ajustent dynamiquement le taux d'apprentissage pour chaque paramètre. Adam est l'un des plus populaires car il combine momentum et taux d'apprentissage adaptatif.

````python
w = w - alpha * (dLoss/dw)
# w = poids, alpha = taux d'apprentissage
````

#### Analogie : une balle qui roule sur une colline

Imagine que tu fais rouler une balle sur une colline pour atteindre le point le plus bas (la vallée), qui représente la perte minimale.

- Descente de gradient — prudente mais lente : tu prends de petits pas calculés, tout droit vers le bas.
- SGD — rapide mais chaotique : tu ne regardes qu'une petite partie de la colline avant de bouger, ce qui te fait parfois trébucher.
- Momentum — un coup de pouce qui roule : tu commences doucement puis prends de la vitesse comme une boule de neige.
- Adam — un GPS intelligent : il ajuste automatiquement ta vitesse et ta direction selon le terrain.

#### Choisir le bon optimiseur

- Petits datasets → SGD ou Momentum fonctionnent bien.
- Réseaux profonds et gros datasets → Adam ou RMSprop sont préférés.
- Données creuses (sparse) → AdaGrad peut être utile.

## La descente de gradient : le processus d'apprentissage

La descente de gradient est le processus qui permet de trouver les meilleurs poids et biais pour le réseau. Elle fonctionne ainsi :

- 1. Commencer avec des poids et des biais aléatoires.
- 2. Calculer la perte (à quel point les prédictions sont fausses) avec les poids et biais actuels.
- 3. Ajuster les poids et biais pour réduire la perte, en calculant le gradient de la perte par rapport à chaque paramètre.
- 4. Mettre à jour les poids et biais avec la formule w = w - α × (∂Perte/∂w).
- 5. Répéter les étapes 2 à 4 jusqu'à ce que la perte soit minimisée.

#### Exemple : trouver le fond d'une vallée

Imagine que tu es dans une vallée et que ton but est de trouver le point le plus bas :

- Regarde autour de toi pour voir quelle direction descend.
- Fais un petit pas dans cette direction.
- Répète jusqu'à atteindre le fond.

Dans un réseau de neurones : la vallée représente la fonction de perte, et le point le plus bas représente les meilleurs poids et biais.

## Tout assembler

- Fonction de perte : mesure l'erreur entre prédictions et valeurs réelles.
- Optimiseur : détermine comment mettre à jour les poids et biais pour minimiser la perte.
- Descente de gradient : le processus itératif de mise à jour des poids et biais.

#### Exemple : entraîner un réseau de neurones

- Propagation avant : le réseau fait des prédictions avec les poids et biais actuels.
- Calcul de la perte : la fonction de perte calcule l'erreur.
- Rétropropagation : les gradients sont calculés grâce à la règle de dérivation en chaîne.
- Mise à jour des paramètres : l'optimiseur met à jour les poids et biais par descente de gradient.
- Répétition : le processus se répète jusqu'à ce que la performance du modèle soit satisfaisante.

Ces trois composants travaillent ensemble pour permettre au réseau d'apprendre à partir des données, de faire de meilleures prédictions et de s'améliorer au fil du temps.

## Qu'est-ce que l'overfitting ?

L'un des plus grands défis de l'entraînement d'un réseau de neurones est l'overfitting (surapprentissage). Il se produit quand le modèle apprend trop bien les données d'entraînement, y compris leur bruit et leurs valeurs aberrantes, et qu'il performe mal sur des données nouvelles, jamais vues.

#### Pourquoi l'overfitting se produit-il ?

L'overfitting survient quand le modèle devient trop complexe et se met à mémoriser les données d'entraînement plutôt qu'à apprendre des motifs généraux. C'est particulièrement fréquent quand :

- Le modèle a trop de paramètres (trop de couches ou de neurones).
- Le dataset d'entraînement est trop petit.

#### Exemple : mémoriser vs. comprendre

Imagine que tu révises pour un examen de maths :

- Si tu mémorises les réponses à des exercices précis, tu réussiras peut-être ces exercices-là, mais tu échoueras sur des nouveaux. C'est de l'overfitting.
- Si tu comprends les concepts, tu peux résoudre n'importe quel exercice, même ceux que tu n'as jamais vus. C'est ce qu'on attend d'un réseau de neurones.

#### Signes d'overfitting

- L'accuracy du modèle sur les données d'entraînement est bien plus élevée que sur les données de test.
- Le modèle performe mal sur des données nouvelles, jamais vues.

> ⚠️ **Important** : La régularisation ajoute des contraintes au modèle, l'incitant à apprendre des motifs essentiels plutôt qu'à mémoriser les données d'entraînement. Sans régularisation, un modèle peut devenir trop complexe et s'ajuster au bruit plutôt qu'aux motifs réellement pertinents.

## Techniques de régularisation courantes

### A. Early stopping

Plutôt que d'entraîner le modèle pour un nombre fixe d'epochs, l'early stopping surveille la perte de validation. Si la perte cesse de s'améliorer (ou commence à augmenter), l'entraînement s'arrête, ce qui évite l'overfitting.

> 💡 **Astuce** : Imagine que tu révises pour un examen : trop peu, tu n'apprends pas assez (underfitting) ; trop, tu mémorises des détails inutiles pour un nouveau test (overfitting). L'early stopping garantit que tu arrêtes de réviser au bon moment.

### B. Data augmentation

Plutôt que de modifier le modèle, la data augmentation augmente la diversité des données d'entraînement — particulièrement utile en vision par ordinateur, où les images peuvent être légèrement pivotées, retournées ou zoomées pour créer des variations.

Exemple : si tu entraînes un modèle à reconnaître des chats, retourner ou modifier la luminosité des photos de chats crée de nouveaux exemples d'entraînement, rendant le modèle plus adaptable.

### C. Batch normalization

La batch normalization standardise l'entrée de chaque couche, garantissant un apprentissage stable. Elle réduit la dépendance à l'initialisation des poids et accélère la convergence.

Analogie : imagine courir un marathon. Si tu pars trop vite, tu t'épuises rapidement. La batch normalization garantit un rythme de course constant, évitant des variations brutales dans l'apprentissage.

## Le dropout : désactiver aléatoirement des neurones

Le dropout est une technique où, pendant l'entraînement, des neurones aléatoires du réseau sont « désactivés » avec une certaine probabilité. Cela empêche le réseau de trop dépendre de neurones spécifiques et l'encourage à apprendre des features plus robustes.

#### Comment fonctionne le dropout

- Pendant l'entraînement, chaque neurone a une probabilité p d'être désactivé. Si p = 0.5, chaque neurone a 50 % de chances d'être « drop out ».
- Les neurones désactivés ne contribuent ni à la propagation avant ni à la rétropropagation.
- Pendant les tests, tous les neurones sont actifs, mais leurs sorties sont mises à l'échelle pour compenser le dropout.

#### Pourquoi le dropout fonctionne

Le dropout force le réseau à apprendre des représentations redondantes. En désactivant aléatoirement des neurones, le réseau ne peut pas se reposer sur un seul neurone et doit distribuer l'apprentissage sur l'ensemble d'entre eux — comme dans un travail de groupe : si une seule personne fait tout le travail, le groupe échoue en son absence ; si tout le monde contribue également, le groupe peut faire face à n'importe quel défi.

````python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),  # Dropout avec 50% de probabilité
    layers.Dense(10, activation='softmax')
])
````

Dropout(0.5) désactive aléatoirement 50 % des neurones de cette couche pendant l'entraînement, forçant le réseau à apprendre des features plus robustes.

## La régularisation L2 : pénaliser les grands poids

La régularisation L2 (aussi appelée weight decay) ajoute une pénalité pour les grands poids à la fonction de perte. Cela encourage le réseau à utiliser des poids plus petits, ce qui réduit l'overfitting.

```python
Nouvelle_Perte = Perte_Originale + lambda * somme(w_i ** 2)
# lambda = force de la régularisation (hyperparamètre)
# w_i = poids du réseau
```

Pendant l'entraînement, l'optimiseur essaie de minimiser à la fois la perte originale et le terme de pénalité, ce qui décourage le réseau de trop dépendre d'une seule feature.

Analogie : si tu écris un essai avec trop de mots compliqués, il peut sembler impressionnant mais devenir difficile à comprendre — comme un réseau avec de grands poids. Un langage simple et clair est plus facile à comprendre, exactement ce que la régularisation L2 encourage.

````python
from tensorflow.keras import layers, models, regularizers

model = models.Sequential([
    layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    layers.Dense(10, activation='softmax')
])
````

## Le problème : classifier CIFAR-10 sans overfitting

Le dataset CIFAR-10 contient 60 000 images réparties en 10 classes (avions, voitures, oiseaux…). L'objectif est de construire un réseau de neurones capable de classifier ces images sans overfitting. Dans ce tutoriel, on utilise des techniques de régularisation — dropout, régularisation L2 et early stopping — pour éviter le surapprentissage.

## Charger et prétraiter les données

````python
import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout, Conv2D, MaxPooling2D, BatchNormalization
````

````python
# Charger le dataset CIFAR-10
(X_train, y_train), (X_test, y_test) = cifar10.load_data()

# Normaliser les valeurs de pixels entre 0 et 1
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding des labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
````

- Le dataset CIFAR-10 comporte 50 000 images d'entraînement et 10 000 images de test.
- Chaque image est en couleur (RGB), de taille 32x32.
- Les valeurs de pixels sont normalisées pour améliorer l'efficacité de l'entraînement.
- Les labels sont encodés en one-hot pour être compatibles avec la categorical crossentropy.

## Construire le modèle avec régularisation

On construit un réseau de neurones convolutif (CNN) avec :

- Couches convolutives : extraient les features des images.
- Batch normalization : normalise les sorties de chaque couche pour stabiliser l'entraînement.
- Max Pooling : réduit les dimensions spatiales des cartes de features.
- Dropout : désactive aléatoirement des neurones pendant l'entraînement pour éviter l'overfitting.
- Couches denses : couches entièrement connectées pour la classification finale.

````python
model = Sequential([
    # Premier bloc convolutif
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.2),

    # Second bloc convolutif
    Conv2D(64, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.3),

    # Couches entièrement connectées
    Flatten(),
    Dense(128, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.01)),
    Dropout(0.4),
    Dense(10, activation='softmax')  # Couche de sortie
])
````

> 💡 **À retenir** : Dropout désactive aléatoirement des neurones pour éviter l'overfitting ; la régularisation L2 pénalise les grands poids pour encourager des modèles plus simples ; la batch normalization normalise les sorties de chaque couche pour stabiliser l'entraînement.

## Compiler, entraîner et évaluer

#### Compiler le modèle

````python
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
````

#### Entraîner avec early stopping

On entraîne le modèle pour 50 epochs avec un batch size de 64. Pour éviter l'overfitting, on utilise l'early stopping, qui arrête l'entraînement si la perte de validation ne s'améliore plus pendant 5 epochs consécutives.

````python
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = model.fit(X_train, y_train,
                     validation_split=0.2,
                     epochs=50,
                     batch_size=64,
                     callbacks=[early_stopping])
````

#### Évaluer sur les données de test

````python
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_acc:.2f}")
````

> ⚠️ **Important** : L'accuracy de test donne une idée de la capacité du modèle à généraliser sur des données nouvelles. Si l'accuracy d'entraînement est nettement supérieure à celle du test, c'est le signe d'un overfitting résiduel — essaie d'augmenter le taux de dropout ou la force de la régularisation L2.
