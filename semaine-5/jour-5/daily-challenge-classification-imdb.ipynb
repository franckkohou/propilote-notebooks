## II. Les étapes

### 1. Prétraiter

Un réseau ne peut pas lire des listes d'entiers de longueurs variables : on les transforme en vecteurs. Avec l'encodage one-hot, la séquence [3, 5] devient un vecteur de 10 000 zéros, sauf aux indices 3 et 5 qui valent 1. Une couche Dense peut alors traiter ces vecteurs.

```python
import numpy as np

def vectorize_sequences(sequences, dimension=10000):
    results = np.zeros((len(sequences), dimension))
    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.   # met à 1 les indices présents
    return results
```
