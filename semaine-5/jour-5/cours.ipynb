## Les réflexes d'un bon projet de deep learning

- Soigner le pipeline de données : si les données sont mal lues ou mal étiquetées, aucun modèle ne rattrapera les performances.
- Garder un jeu de validation fixe et intact : on n'augmente que les données d'entraînement.
- Suivre la perte ET l'exactitude : l'exactitude trompe quand les classes sont déséquilibrées.
- Détecter le surapprentissage : quand la courbe de validation s'écarte de celle d'entraînement, renforcer l'augmentation, ajouter du dropout, réduire le modèle ou arrêter plus tôt (EarlyStopping).
- Comparer à une référence (ablation) : même architecture avec et sans augmentation, pour mesurer ce qui aide vraiment.
- Sauvegarder le modèle et sa configuration pour pouvoir reproduire les résultats.
