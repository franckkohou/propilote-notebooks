> Implémente en Python un réseau de neurones simple qui effectue une propagation avant puis une rétropropagation pour un problème de régression (prédire le score d'examen à partir des heures d'étude).
>
> - Exécute le code et observe comment les poids et le biais évoluent.
> - Explique pourquoi la mise à jour des poids par descente de gradient réduit l'erreur.
> - Modifie les poids initiaux ou le taux d'apprentissage et observe l'effet sur l'apprentissage.
>
> ```python
> import numpy as np
>
> # Initialisation des données d'entrée (features)
> x = np.array([4, 80])  # 4 heures étudiées, note précédente : 80
>
> # Initialisation des poids et du biais
> w = np.array([0.6, 0.3])  # Poids initiaux
> b = 10  # Biais initial
>
> # Propagation avant
> def forward_propagation(x, w, b):
>     z = np.dot(x, w) + b  # Somme pondérée
>     return z  # Activation linéaire (pas de ReLU, c'est une régression)
>
> # Calcul de la prédiction
> y_pred = forward_propagation(x, w, b)
> y_true = 85  # Score réel de l'examen
>
> # Calcul de la perte (Mean Squared Error)
> loss = 0.5 * (y_true - y_pred) ** 2
>
> # Calcul des gradients
> grad_w = -(y_true - y_pred) * x  # Dérivées partielles par rapport aux poids
> grad_b = -(y_true - y_pred)      # Dérivée partielle par rapport au biais
>
> # Mise à jour des poids et du biais
> learning_rate = 0.01
> w_new = w - learning_rate * grad_w
> b_new = b - learning_rate * grad_b
>
> # Affichage des résultats
> print("Initial Prediction:", y_pred)
> print("Loss:", loss)
> print("Updated Weights:", w_new)
> print("Updated Bias:", b_new)
> ```
