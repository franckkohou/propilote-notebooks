# 1. Introduction au Machine Learning

*Définition intuitive, intérêt pour un data analyst et grands domaines d'application*

**Objectifs**

- Comprendre intuitivement comment une machine « apprend »

## Qu'est-ce que le Machine Learning ?

Le Machine Learning est une méthode qui permet à un ordinateur d'apprendre à partir de données pour prendre des décisions ou faire des prédictions, sans être explicitement programmé pour chaque tâche.

#### Une analogie : apprendre à un enfant à reconnaître des fruits

Imagine que tu apprennes à un enfant à distinguer différents fruits. Tu lui montres plusieurs fruits en les nommant. Avec le temps, l'enfant apprend à identifier chaque fruit à partir de ses caractéristiques (forme, couleur, taille). Le Machine Learning fonctionne de façon similaire : on fournit à l'ordinateur des données (ici, des informations sur des fruits), et il apprend à reconnaître des motifs ou des caractéristiques (comme la forme et la couleur d'une pomme) pour prendre une décision (identifier le fruit).

Tout comme un enfant s'améliore avec la pratique, un modèle de Machine Learning améliore ses prédictions à mesure qu'il traite davantage de données : plus il reçoit d'exemples, plus il devient précis.

> 💡 **Astuce** : La grande force du Machine Learning est sa capacité à s'adapter à de nouvelles données jamais vues. Une fois entraîné, l'algorithme peut appliquer ce qu'il a appris à des données inédites : prédire un comportement client, reconnaître de la parole, diagnostiquer une maladie…

> ⚠️ **Attention** : Comme les humains, les modèles de Machine Learning ne sont pas parfaits. Ils peuvent se tromper, en particulier si les données sur lesquelles ils apprennent sont peu diversifiées ou biaisées. Un entraînement soigné et un affinage régulier permettent toutefois d'atteindre une précision impressionnante.

## Pourquoi le ML pour un data analyst ?

Les compétences principales d'un data analyst sont en général le nettoyage de données, la visualisation et l'analyse statistique. Mais dans certains secteurs ou postes, le ML devient un atout décisif — par exemple si tu travailles dans une entreprise tech qui manipule de gros volumes de données et fait de la modélisation prédictive.

Même quand le ML n'est pas une exigence stricte du poste, en avoir la maîtrise te différencie des autres candidats : cela montre que tu as des compétences plus larges et que tu peux prendre en charge des tâches plus complexes.

## Les usages du Machine Learning

Le Machine Learning est utilisé dans un très grand nombre de secteurs. En voici quelques exemples clés :

- Analyse prédictive : prévision des tendances financières, de la demande en retail, de la météo.
- Reconnaissance d'image et de la parole : reconnaissance faciale, diagnostic médical par imagerie, assistants vocaux (Siri, Alexa, Google Assistant).
- Traitement du langage naturel (NLP) : traduction automatique, analyse de sentiment, chatbots.
- Systèmes de recommandation : Amazon, Netflix, Spotify analysent ton comportement passé pour te recommander des produits, films ou musiques.
- Véhicules autonomes : décisions en temps réel (détection d'objets, anticipation du comportement des piétons, choix d'itinéraire).
- Santé : détection précoce de maladies, médecine personnalisée, automatisation de tests diagnostiques.
- Détection de fraude : repérage de transactions bancaires suspectes en temps réel.
- Marketing personnalisé, maintenance prédictive en industrie, jeux vidéo (PNJ plus intelligents), agriculture de précision, protection de l'environnement, éducation personnalisée, recrutement (RH), robotique.

> 💡 **À retenir** : Retiens surtout la logique commune à tous ces usages : on donne des données au modèle, il en extrait des motifs, et il s'en sert pour prédire ou décider sur de nouvelles données.

# 2. Les types de Machine Learning

*Apprentissage supervisé, non supervisé et par renforcement — comment les reconnaître*

**Objectifs**

- Distinguer apprentissage supervisé, non supervisé et par renforcement
- Reconnaître à quel type appartient un scénario donné
- Comprendre les notions de classification, régression et clustering

## Apprentissage supervisé

En apprentissage supervisé, l'algorithme apprend à partir d'un jeu de données étiqueté : on connaît à l'avance la sortie associée à chaque entrée.

Exemple : un jeu de données d'images, où l'objectif est de classer chaque image comme contenant un chat ou non. Chaque image est étiquetée « chat » ou « pas chat ». C'est un exemple classique de classification binaire.

Le but est de prédire la sortie pour de nouvelles données. Une fois le modèle entraîné — un algorithme capable de dire s'il y a un chat sur une image —, on peut l'utiliser sur de nouvelles images et lui demander si elles contiennent un chat ou non.

## Apprentissage non supervisé

L'apprentissage non supervisé entraîne l'algorithme sur des données ni classées ni étiquetées. Le système essaie de découvrir seul les motifs et structures des données, en regroupant les points de données qui se ressemblent : c'est ce qu'on appelle le clustering.

Exemple : trier un jeu de cartes sans en connaître les règles. Le système pourrait les trier par numéro, par couleur ou par famille, selon les motifs qu'il observe.

## Apprentissage par renforcement

L'apprentissage par renforcement consiste à entreprendre les actions les plus adaptées pour maximiser une récompense dans une situation donnée. C'est un type de ML où un agent apprend à prendre des décisions en effectuant des actions et en recevant un retour sous forme de récompenses ou de pénalités.

Ce processus ressemble à l'entraînement d'un animal domestique avec des friandises pour les bons comportements et de légères réprimandes pour les mauvais.

Exemple : un système de filtrage des emails traité comme un agent qui classe les emails (spam / pas spam). Au départ, il ne sait presque rien et fait des suppositions. Le retour de l'utilisateur (correction ou confirmation) constitue la récompense ou la pénalité, et le système ajuste progressivement ses critères de classification.

## Comment reconnaître le type de ML utilisé ?

### A. Apprentissage supervisé

- Présence de données étiquetées : à chaque entrée correspond une sortie connue.
- L'algorithme reçoit un retour direct sur ses prédictions (correcte ou non).
- L'objectif est souvent de prédire une valeur numérique (régression) ou une catégorie (classification).

### B. Apprentissage non supervisé

- Les données arrivent sans étiquette : l'algorithme ne connaît pas la catégorie à l'avance.
- L'accent est mis sur la découverte de motifs, structures ou relations.
- Objectifs courants : clustering (regroupement de points similaires) ou association (règles décrivant de grandes parties des données).

### C. Apprentissage par renforcement

- Le modèle apprend en interagissant avec un environnement et en recevant des récompenses ou pénalités.
- Accent mis sur l'exploration (essayer de nouvelles choses) et l'exploitation (utiliser les stratégies déjà connues).
- Le modèle prend des décisions (actions) et apprend des conséquences de ces actions.

# 3. Quel est le processus d'un modèle de ML ?

*De la compréhension du problème à l'évaluation : les étapes clés, comme une recette de cuisine*

**Objectifs**

- Comprendre les grandes étapes de la construction d'un modèle de ML
- Distinguer entraînement (training) et test
- Savoir sélectionner des features pertinentes
- Comprendre les métriques d'évaluation : accuracy, precision, recall, F1-score
- Comprendre les fonctions de perte (loss functions) usuelles

## Comprendre le problème et collecter les données

Développer un modèle de Machine Learning ressemble à la préparation d'un plat par un chef : il faut les bons ingrédients (les données), la bonne recette (l'algorithme) et le bon procédé de cuisson (l'entraînement et le test) pour obtenir un résultat réussi (un modèle performant).

#### Comprendre le problème

Comme un chef doit savoir quel plat il prépare, il faut d'abord définir clairement le problème à résoudre avec le ML : prédire une tendance future, classer des données en catégories, etc.

#### Collecte des données

C'est l'étape où l'on rassemble les ingrédients de la recette. La qualité et la quantité des données ont un impact énorme sur la performance du modèle final.

On sépare toujours le jeu de données en deux parties : une partie pour l'entraînement (le modèle apprend) et une partie pour le test (on vérifie sa précision sur des données jamais vues).

- Entraînement (training) : processus par lequel le modèle apprend à partir des données, par exemple en lui donnant un ensemble d'images étiquetées pour qu'il reconnaisse un chat.
- Test : évaluation de la performance du modèle sur des données nouvelles, jamais vues pendant l'entraînement.

## Sélection des features et choix du modèle

On décide ensuite quelles caractéristiques (features/variables) des données sont pertinentes pour le problème. Comme un chef sélectionne des ingrédients qui se marient bien, on sélectionne les features les plus prédictives pour le modèle.

Exemple : pour prédire le prix d'une maison, on dispose de données comme la surface, le nombre de chambres et de salles de bains, l'âge du bien, la localisation, la proximité des écoles et des transports, des rénovations récentes, le taux de criminalité local, les conditions de marché…

- On commence par identifier les features réputées importantes (surface, nombre de chambres, localisation).
- On analyse ensuite les corrélations, par exemple entre la surface et le prix.
- Si deux features sont fortement corrélées (surface et nombre de chambres), on peut n'en garder qu'une pour éviter la redondance.
- Certaines features pertinentes en théorie peuvent être écartées si les données sont peu fiables (ex. rénovations mal renseignées).

### Choix du modèle

Choisir un modèle, c'est comme choisir une recette : tout dépend du résultat souhaité. Selon le problème, on choisit un algorithme adapté — régression linéaire simple, réseau de neurones complexe, arbres de décision, etc. Nous détaillerons chaque famille de modèles dans les jours suivants.

## Entraînement et évaluation du modèle

Pendant l'entraînement, le modèle apprend la relation entre les features et la variable cible en ajustant ses paramètres pour coller aux données — comme on suit une recette en ajustant les quantités.

Après l'entraînement, on évalue la performance sur un jeu de test séparé pour comprendre comment le modèle se comportera sur des données inédites.

#### Vrais/faux positifs et négatifs

- Vrai Positif (TP) : le modèle prédit correctement la classe positive (ex. il détecte correctement une maladie chez une personne malade).
- Vrai Négatif (TN) : le modèle prédit correctement la classe négative (il identifie correctement une personne non malade).
- Faux Positif (FP) : le modèle prédit à tort la classe positive — « erreur de Type I » (il dit malade alors que non).
- Faux Négatif (FN) : le modèle prédit à tort la classe négative — « erreur de Type II » (il ne détecte pas une maladie réellement présente).

```python
Accuracy  = (TP + TN) / (TP + TN + FP + FN)
Precision = TP / (TP + FP)
Recall    = TP / (TP + FN)
F1-Score  = 2 * (Precision * Recall) / (Precision + Recall)
```

- Accuracy : proportion de prédictions correctes. Utile sur des jeux équilibrés, mais trompeuse sur des classes déséquilibrées (ex. détection de fraude, où les cas positifs sont rares).
- Precision : proportion de vrais positifs parmi toutes les prédictions positives. Important quand un faux positif coûte cher (ex. classer un email légitime comme spam).
- Recall (sensibilité) : proportion des cas positifs réels correctement identifiés. Crucial quand un faux négatif coûte cher (ex. rater un cancer).
- F1-Score : moyenne harmonique de precision et recall, utile pour trouver un équilibre entre les deux, en particulier sur des classes déséquilibrées.

#### Validation croisée et courbes d'apprentissage

- K-Fold Cross-Validation : les données sont divisées en K sous-ensembles ; chacun sert une fois de validation pendant que le modèle est entraîné sur les K-1 autres. Répété K fois.
- Stratified K-Fold : identique, mais en conservant la proportion de chaque classe dans chaque fold — utile pour les jeux déséquilibrés.
- Courbes d'apprentissage : comparent la performance sur l'entraînement et la validation selon la quantité de données ou la complexité du modèle.

> ⚠️ **Attention** : Sous-apprentissage (underfitting) : scores faibles en entraînement et en validation — le modèle est trop simple. Surapprentissage (overfitting) : score élevé en entraînement mais faible en validation — le modèle colle trop au bruit des données d'entraînement. L'objectif est un bon compromis (« good fit »), où les deux scores convergent vers une valeur élevée.

## Fonctions de perte (loss functions)

Une fonction de perte (ou fonction de coût) mesure à quel point les prédictions d'un modèle correspondent aux valeurs réelles. Elle quantifie l'erreur entre la sortie prédite et la sortie réelle. L'objectif de l'entraînement est de minimiser cette erreur.

- Mean Squared Error (MSE) : utilisée pour la régression (ex. prédire le prix d'une maison). Mesure la moyenne des écarts au carré entre prédictions et valeurs réelles.
- Categorical Crossentropy : utilisée pour la classification multi-classes (ex. reconnaître des chiffres manuscrits). Mesure l'écart entre les probabilités prédites et les vraies étiquettes.
- Binary Crossentropy : utilisée pour la classification binaire (ex. détecter une transaction frauduleuse).

```python
MSE = (1/n) * Σ (y_i - ŷ_i)²
```

````python
# Exemple : prédiction du prix d'une maison
# Prix prédit : 300 000 $ — Prix réel : 350 000 $
MSE = (300000 - 350000) ** 2
# => 2 500 000 000
# Plus l'erreur est grande, plus la perte est élevée.
````

## Pourquoi le ML est important pour la Gen AI

Le Machine Learning est le socle de l'IA générative : il permet aux modèles d'analyser d'immenses quantités de données, d'apprendre des motifs complexes, et de générer des contenus entièrement nouveaux (texte, images, musique, voire code).

Utiliser un modèle génératif pré-entraîné demande peu de connaissances en ML, mais comprendre les concepts fondamentaux du ML apporte plusieurs avantages :

- Fine-tuning : affiner un modèle existant pour des tâches spécifiques en l'entraînant sur des jeux de données spécialisés.
- Contrôle des sorties : mieux guider le comportement du modèle, améliorer la cohérence, réduire les biais.
- Construction de solutions sur-mesure : créer des systèmes d'IA adaptés à des besoins métier ou de recherche précis, plutôt que de dépendre uniquement de modèles « sur étagère ».

Les modèles génératifs pilotés par le ML apprennent à partir de vastes jeux de données et appliquent des méthodes probabilistes pour générer du nouveau contenu qui imite les données d'entraînement :

- GPT-4 (génération de texte) : entraîné sur d'immenses corpus de texte, il apprend la structure du langage et le contexte pour générer des réponses proches de celles d'un humain.
- DALL·E (génération d'image) : transforme un prompt textuel en image réaliste grâce au deep learning.
- GANs (IA créative) : génèrent des images réalistes, des compositions musicales, voire des deepfakes.

> 💡 **Astuce** : En maîtrisant les fondamentaux du ML, tu ne te contentes plus d'utiliser l'IA générative : tu peux repousser ses limites, l'affiner pour des tâches précises, et débloquer de nouvelles possibilités créatives.

## Cas d'usage du ML pour la Gen AI

Le Machine Learning joue un rôle central dans de nombreuses applications de l'IA générative :

- Génération de texte et chatbots : les modèles de type GPT (Generative Pretrained Transformer) créent du texte proche de celui d'un humain, pour des chatbots, la génération de contenu ou la narration automatisée. Exemple : automatiser des réponses de service client, générer des articles de blog ou des descriptions produit.
- Génération d'image et de vidéo : des modèles entraînés sur d'immenses jeux d'images/vidéos, souvent via des GANs (Generative Adversarial Networks), génèrent des visuels nouveaux et réalistes. Exemple : MidJourney génère des illustrations à partir de descriptions textuelles.
- Synthèse audio et musicale : le ML permet de composer de la musique ou de générer de la parole réaliste (ex. Google WaveNet pour le text-to-speech).
- Génération et automatisation de code : des modèles entraînés sur d'immenses corpus de code comprennent la syntaxe et le contexte pour suggérer ou écrire du code. Exemple : GitHub Copilot, propulsé par OpenAI Codex, complète du code à partir de commentaires ou de fragments.
- Augmentation de données et données synthétiques : générer des données synthétiques réalistes (ex. images médicales) pour entraîner des modèles sans compromettre la confidentialité des patients, quand les données réelles sont rares ou sensibles.
