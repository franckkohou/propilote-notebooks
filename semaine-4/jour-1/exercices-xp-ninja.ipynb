## I. Implémentation des métriques d'évaluation

#### Objectif

Implémenter en Python des fonctions calculant l'Accuracy, la Precision, le Recall et le F1-Score à partir de zéro (sans utiliser les fonctions toutes faites de scikit-learn pour ces calculs).

#### Jeu de données

Utilise n'importe quel jeu de données de classification binaire, par exemple le dataset Breast Cancer Wisconsin (Diagnostic) disponible dans scikit-learn, ou tout autre dataset de classification binaire de ton choix.

#### Tâches

- Sépare le jeu de données en un ensemble d'entraînement et un ensemble de test.
- Entraîne un modèle de classification simple (comme une régression logistique) sur l'ensemble d'entraînement.
- Fais des prédictions sur l'ensemble de test.
- Écris des fonctions calculant l'Accuracy, la Precision, le Recall et le F1-Score à partir des composantes de la matrice de confusion (TP, TN, FP, FN).
- Évalue ton modèle avec ces métriques et interprète les résultats.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

def confusion_counts(y_true, y_pred):
    tp = sum((yt == 1 and yp == 1) for yt, yp in zip(y_true, y_pred))
    tn = sum((yt == 0 and yp == 0) for yt, yp in zip(y_true, y_pred))
    fp = sum((yt == 0 and yp == 1) for yt, yp in zip(y_true, y_pred))
    fn = sum((yt == 1 and yp == 0) for yt, yp in zip(y_true, y_pred))
    return tp, tn, fp, fn

def accuracy(tp, tn, fp, fn):
    return (tp + tn) / (tp + tn + fp + fn)

def precision(tp, fp):
    return tp / (tp + fp) if (tp + fp) else 0

def recall(tp, fn):
    return tp / (tp + fn) if (tp + fn) else 0

def f1_score(prec, rec):
    return 2 * prec * rec / (prec + rec) if (prec + rec) else 0

tp, tn, fp, fn = confusion_counts(y_test, y_pred)
prec = precision(tp, fp)
rec = recall(tp, fn)
print("Accuracy :", accuracy(tp, tn, fp, fn))
print("Precision:", prec)
print("Recall   :", rec)
print("F1-Score :", f1_score(prec, rec))
```
