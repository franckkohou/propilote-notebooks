## Les p-values

Les statistiques inférentielles permettent de faire des prédictions ou des inférences sur une population à partir d'un échantillon de données. Elles sont une pierre angulaire de la data science et du Machine Learning, car elles permettent de tirer des conclusions, tester des hypothèses et prendre des décisions dans l'incertitude.

La p-value est la probabilité d'observer les données de l'échantillon (ou quelque chose de plus extrême), en supposant que l'hypothèse nulle soit vraie. Elle permet de déterminer si les preuves apportées par les données sont assez fortes pour rejeter l'hypothèse nulle. Une p-value faible (généralement < 0,05) suggère que les données observées sont peu probables sous l'hypothèse nulle, ce qui conduit à son rejet.

- Analyse de données : la p-value aide à déterminer la force des preuves contre l'hypothèse nulle et à évaluer l'importance de chaque prédicteur dans un modèle.
- Machine Learning : dans les modèles de régression, les p-values servent à évaluer la pertinence de chaque feature. Les features à faible p-value sont souvent conservées, celles à p-value élevée peuvent être retirées.
- Sélection de features : les p-values aident à sélectionner les features significatives qui améliorent la performance du modèle.

Exemple : on veut prédire le score d'examen des étudiants selon leur temps de révision. Si la p-value pour la feature « heures de révision » est de 0,03, cela signifie qu'il y a 3% de chances d'observer une relation aussi forte entre temps de révision et score si l'hypothèse nulle (pas de relation) était vraie. Comme 0,03 < 0,05, on rejette l'hypothèse nulle : le temps de révision est un prédicteur statistiquement significatif du score.

## La F-value

La F-value est une statistique de test qui permet de déterminer la significativité globale d'un modèle de régression. C'est un ratio de variances qui compare la variance expliquée (due aux prédicteurs du modèle) à la variance inexpliquée (due au bruit aléatoire). Une F-value élevée indique que le modèle explique une plus grande part de la variation de la variable cible.

- Significativité du modèle : la F-value teste si au moins un des prédicteurs contribue significativement à expliquer la variable cible.
- Évaluation globale du modèle : contrairement aux p-values, qui testent des prédicteurs individuels, la F-value teste l'efficacité globale du modèle.
- Comparaison à la p-value : la F-value est souvent accompagnée d'une p-value pour évaluer si le modèle global est statistiquement significatif.

Exemple : pour un modèle prédisant le salaire des employés à partir de l'expérience, du niveau d'éducation et du département, supposons une F-value de 15,6 avec une p-value de 0,001. Le modèle global est statistiquement significatif : au moins un des prédicteurs a une relation significative avec le salaire.

## Le test d'hypothèse

Le test d'hypothèse est une méthode statistique utilisée pour prendre des décisions ou faire des inférences sur un paramètre de population à partir de données d'échantillon. Il s'appuie sur deux hypothèses concurrentes :

- Hypothèse nulle (H₀) : affirme qu'il n'y a pas d'effet ou de différence (ex. « la moyenne du groupe A est égale à celle du groupe B »). Elle contient toujours une égalité (=, ≤, ≥).
- Hypothèse alternative (H₁) : contredit l'hypothèse nulle (ex. « la moyenne du groupe A est différente de celle du groupe B »). Elle définit l'effet qu'on cherche à détecter (≠, >, <).
- Statistique de test : une valeur calculée (t-statistique, z-score…) qui mesure à quel point le résultat de l'échantillon s'éloigne de H₀.
- Seuil de significativité (α) : le seuil au-delà duquel on rejette H₀ (généralement 0,05). Il contrôle le taux d'erreur de Type I (faux positifs).

- Erreur de Type I : rejeter H₀ alors qu'elle est vraie (faux positif).
- Erreur de Type II : ne pas rejeter H₀ alors qu'elle est fausse (faux négatif).

#### Les étapes du test d'hypothèse

- Définir H₀ et H₁ : par exemple H0: β=0 (pas d'effet) contre H1: β≠0 (effet existant).
- Choisir α : généralement 0,05 (5% de risque d'erreur de Type I).
- Sélectionner le test : t-test pour les coefficients de régression (petits échantillons, variance inconnue) ou z-test (grands échantillons).
- Calculer la statistique de test : t = (β̂ - 0) / SE(β̂), où β̂ est le coefficient estimé et SE son erreur standard.
- Déterminer la p-value : en comparant la statistique de test à sa distribution d'échantillonnage.
- Conclure : si p-value < α, on rejette H₀. Sinon, on ne rejette pas H₀.

Exemple en Machine Learning : pour un modèle de régression linéaire prédisant le prix d'une maison, on veut tester si le nombre de chambres a un effet significatif. H₀ : le coefficient du nombre de chambres est nul. H₁ : il n'est pas nul. Un t-test sur ce coefficient permet de trancher.

## Les intervalles de confiance

Un intervalle de confiance (IC) est une plage statistique, associée à une probabilité, susceptible de contenir la vraie valeur d'un paramètre de population (comme une moyenne ou une proportion). Le niveau de confiance le plus courant est 95% : si l'on répétait l'expérience de nombreuses fois, environ 95% des intervalles calculés contiendraient la vraie valeur du paramètre.

- Un intervalle étroit suggère une estimation précise, avec peu de variation dans les données.
- Un intervalle large suggère une estimation moins précise, avec une incertitude considérable.
- Si un intervalle de confiance inclut zéro (pour un coefficient de régression), cela indique que le paramètre n'a peut-être pas d'effet significatif.

Exemple : pour un modèle prédisant les ventes à partir des dépenses publicitaires, l'IC à 95% du coefficient de dépense publicitaire est [200, 500]. On est confiant à 95% que chaque unité supplémentaire de dépense publicitaire augmente les ventes de 200 à 500 unités.

> 💡 **Astuce** : Un intervalle très étroit (ex. [480, 500]) indique une estimation précise. Un intervalle large (ex. [100, 500]) indique une plus grande incertitude, et suggère qu'il faudrait peut-être collecter plus de données ou affiner le modèle.

## Applications aux modèles de ML supervisés

- Régression linéaire : les tests d'hypothèse (t-tests sur les coefficients) évaluent la significativité de chaque feature ; les p-values identifient les features significatives ; les intervalles de confiance indiquent la précision des estimations.
- Régression logistique : les tests d'hypothèse évaluent la significativité des prédicteurs dans les tâches de classification ; les p-values déterminent si une feature affecte significativement la probabilité de la classe cible.
- Arbres de décision et forêts aléatoires : moins directement utilisés, mais l'importance des features peut être testée statistiquement (tests de permutation).
- Support Vector Machines (SVM) : les tests d'hypothèse comparent la performance de différents noyaux ou réglages d'hyperparamètres.

````python
import pandas as pd
import statsmodels.api as sm

data = pd.read_csv('house_prices.csv')

X = data[['Area', 'Bedrooms', 'Bathrooms', 'Stories', 'MainRoad', 'GuestRoom',
           'Basement', 'HotWaterHeater', 'AirConditioning', 'Parking', 'PrefArea']]
y = data['Price']

X = pd.get_dummies(X, drop_first=True)
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())  # coefficients, p-values, intervalles de confiance

print("F-value :", model.fvalue)
print(model.conf_int())  # bornes inférieure/supérieure de chaque coefficient
````

> 💡 **À retenir** : Le test d'hypothèse permet d'évaluer la significativité des relations et des modèles. Les p-values quantifient la force des preuves contre l'hypothèse nulle. Les F-values évaluent la significativité globale d'un modèle de régression. Les intervalles de confiance donnent une plage de valeurs plausibles pour un paramètre de population.

## Le test d'hypothèse en Machine Learning

En Machine Learning, prendre des décisions fondées sur les données — sur les modèles, les features et les résultats métier — est essentiel. Le test d'hypothèse permet de valider si des résultats observés sont statistiquement significatifs ou dus au hasard. L'A/B testing permet quant à lui de comparer deux modèles pour déterminer lequel est le plus performant sur une métrique donnée.

#### Validation de modèles

Le test d'hypothèse joue un rôle crucial pour valider des modèles de ML en comparant différentes métriques de performance (accuracy, F1-score, AUC-ROC). Exemple : tu as deux modèles A et B et veux savoir si la meilleure accuracy de B est statistiquement significative. Le test d'hypothèse permet de tester H0 (les deux modèles ont la même accuracy) et de déterminer si la différence est significative ou due au hasard.

#### Sélection de features

Toutes les features ne contribuent pas également à la performance du modèle. Des tests statistiques (Chi-Square, t-tests) permettent d'évaluer si une feature a un impact significatif sur la variable cible.

#### Décisions métier

Le test d'hypothèse soutient aussi des décisions métier en comparant des modèles sur des métriques comme l'engagement utilisateur, le chiffre d'affaires ou le taux de conversion — c'est le principe même de l'A/B testing.

## T-test ou Z-test : lequel choisir ?

Le t-test et le z-test comparent tous deux des moyennes ou proportions d'échantillon à des valeurs hypothétiques ou entre elles. Le choix dépend de plusieurs facteurs :

| Facteur | Z-test | T-test |
| --- | --- | --- |
| Taille de l'échantillon | Grande (n > 30) | Petite (n ≤ 30) |
| Écart-type de la population | Connu ou grand échantillon (estimation fiable) | Inconnu |

- Z-test : utilisé quand l'échantillon est grand (n > 30), que l'écart-type de la population est connu, et que les données sont approximativement normales. Cas d'usage : comparer des moyennes ou des proportions sur de grands échantillons (ex. CTR entre deux modèles avec 500+ participants par groupe).
- T-test : utilisé quand l'échantillon est petit (n ≤ 30), que l'écart-type de la population est inconnu (cas le plus fréquent en pratique). Le t-test est plus robuste aux écarts à la normalité que le z-test, surtout sur petits échantillons.

- T-test à un échantillon : comparer la moyenne d'un échantillon à une valeur connue.
- T-test à deux échantillons indépendants : comparer les moyennes de deux groupes indépendants.
- T-test apparié : comparer deux groupes liés (ex. mesures avant/après sur le même groupe).

## L'A/B Testing pour comparer des modèles

L'A/B testing (ou split testing) compare deux modèles (A et B) pour déterminer lequel performe le mieux sur une métrique donnée (taux de conversion, click-through rate…).

#### Les étapes d'un A/B test en ML

- Définir les métriques : choisir les critères d'évaluation (accuracy, revenu, rétention…).
- Séparer les données : répartir aléatoirement le jeu de données en groupe contrôle (Modèle A) et groupe traitement (Modèle B).
- Lancer l'expérience : déployer les deux modèles et collecter les données de performance.
- Test statistique : z-test pour les grands échantillons, t-test pour les petits.
- Interpréter les résultats : p-value < 0,05 → on rejette H0 (différence significative). Des intervalles de confiance qui se chevauchent suggèrent l'absence de différence significative.

#### Exemple : comparer le CTR de deux modèles de recommandation

Modèle A (contrôle) : CTR = 12%. Modèle B (traitement) : CTR = 15%. Échantillon : n = 1000 pour chaque modèle.

````python
SE = sqrt( pA*(1-pA)/nA + pB*(1-pB)/nB )
````

```python
# pA = 0.12, pB = 0.15, nA = nB = 1000
SE = sqrt(0.12*0.88/1000 + 0.15*0.85/1000)
SE = sqrt(0.0001056 + 0.0001275)
SE ≈ 0.01526
```

```python
Z = (pA - pB) / SE
Z = (0.12 - 0.15) / 0.01526 ≈ -1.97
```

Pour Z = -1,97, la p-value est d'environ 0,03. Comme 0,03 < 0,05 (seuil α), on rejette l'hypothèse nulle : la différence de CTR entre le Modèle A et le Modèle B est statistiquement significative. Le Modèle B (CTR = 15%) est donc considéré comme meilleur que le Modèle A (CTR = 12%).

> 💡 **Astuce** : Le Z-score mesure à quel point les proportions d'échantillon s'éloignent l'une de l'autre, relativement à la variabilité des données. La p-value indique la probabilité d'observer ces données si l'hypothèse nulle est vraie : plus elle est petite, plus la preuve contre H0 est forte.

## Le bootstrapping

Le bootstrapping est une technique de rééchantillonnage utilisée pour estimer la distribution d'échantillonnage d'une statistique, en tirant de façon répétée des échantillons du jeu de données original, avec remise. Chaque point de donnée peut donc être sélectionné plusieurs fois dans un même rééchantillonnage. C'est particulièrement utile quand le jeu de données est petit ou que sa distribution sous-jacente est inconnue.

- Données limitées : le bootstrapping permet de générer plusieurs jeux de données rééchantillonnés pour simuler des jeux plus grands et évaluer la performance du modèle.
- Stabilité du modèle : il aide à évaluer la sensibilité de la performance du modèle aux variations des données d'entraînement.
- Intervalles de confiance : il permet de calculer des intervalles de confiance pour des métriques de performance, donnant une plage de valeurs plausibles reflétant l'incertitude de l'estimation.

#### Les étapes du bootstrapping

- Rééchantillonner : tirer aléatoirement, avec remise, plusieurs échantillons bootstrap (ex. 1000 rééchantillons) de la même taille que le jeu de données original.
- Calculer la métrique : pour chaque échantillon bootstrap, entraîner le modèle et calculer la métrique souhaitée (accuracy, MSE…).
- Analyser la distribution : calculer la moyenne, l'écart-type ou les percentiles pour comprendre la variabilité et les intervalles de confiance. Un IC à 95% peut être obtenu en prenant les 2,5e et 97,5e percentiles des métriques bootstrapées.

Exemple : sur un petit jeu de données de 100 points, on crée 1000 échantillons bootstrap, on calcule l'accuracy sur chacun, on obtient une moyenne de 88%, et un IC à 95% de [85%, 91%]. Le modèle est donc estimé à 88% d'accuracy, avec une confiance de 95% que la vraie accuracy se situe entre 85% et 91%.

- Évaluation de modèle : le bootstrapping estime la variabilité des métriques de performance, évitant de se fier à un seul jeu de données potentiellement non représentatif.
- Réglage des hyperparamètres : il aide à évaluer comment la performance varie selon les jeux de données, pour identifier des réglages robustes.
- Estimation de l'incertitude : il aide à estimer l'incertitude des prédictions d'un modèle, utile pour des décisions métier basées sur ces sorties.

> ⚠️ **Attention** : Le bootstrapping et la validation croisée sont deux techniques différentes : la validation croisée découpe les données en K folds disjoints pour évaluer la généralisation, tandis que le bootstrapping tire des échantillons avec remise pour estimer la variabilité d'une statistique. Elles répondent à des questions complémentaires, pas identiques.
