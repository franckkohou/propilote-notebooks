## II. Tests d'hypothèse

Test 1 : comparer passagers domestiques vs. internationaux.

- H0 : moyenne des passagers domestiques = moyenne des passagers internationaux.
- H1 : moyenne des passagers domestiques ≠ moyenne des passagers internationaux.
- Méthode : t-test indépendant avec stats.ttest_ind().

Test 2 : tester la significativité de la corrélation entre passagers totaux et vols totaux.

- H0 : pas de corrélation entre passagers totaux et vols totaux (ρ = 0).
- H1 : une corrélation significative existe (ρ ≠ 0).
- Méthode : test de corrélation de Pearson avec stats.pearsonr().
