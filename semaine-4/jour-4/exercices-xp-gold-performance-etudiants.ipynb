## II. Comparaison des scores par genre (t-tests)

Ce que tu vas créer : une analyse des différences de performance entre genres en Mathématiques, Lecture et Écriture, en menant des t-tests pour comparer les scores entre garçons et filles.

- Visualise la distribution des scores (Math, Reading, Writing) par genre, par exemple avec des boxplots ou des histogrammes superposés.
- Pour chaque matière, formule H0 (pas de différence de moyenne entre genres) et H1 (différence de moyenne).
- Réalise un t-test indépendant à deux échantillons entre les scores des filles et des garçons pour chaque matière.

```python
from scipy import stats
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

math_female = df[df["gender"] == "female"]["math score"]
math_male = df[df["gender"] == "male"]["math score"]

t_stat, p_value = stats.ttest_ind(math_female, math_male, equal_var=False)
print(f"T-statistic: {t_stat:.3f}, p-value: {p_value:.4f}")
```

## III. Hypothèses du t-test

Avant de conclure, vérifie les hypothèses sous-jacentes au t-test :

- Normalité : les scores de chaque groupe suivent-ils approximativement une distribution normale ? (test de Shapiro-Wilk, ou inspection visuelle avec un Q-Q plot).
- Homoscédasticité : les variances des deux groupes sont-elles comparables ? (test de Levene).

```python
from scipy.stats import shapiro, levene

print(shapiro(math_female))
print(shapiro(math_male))
print(levene(math_female, math_male))
```

## IV. ANOVA à un facteur

Pour comparer plus de deux groupes à la fois (par exemple le niveau de préparation au test : none / completed), utilise une ANOVA à un facteur (One-Way ANOVA) plutôt qu'une série de t-tests.

```python
from scipy.stats import f_oneway

groups = [df[df["test preparation course"] == g]["math score"] for g in df["test preparation course"].unique()]
f_stat, p_value = f_oneway(*groups)
print(f"F-statistic: {f_stat:.3f}, p-value: {p_value:.4f}")
```

Si l'ANOVA révèle une différence globale significative (p < 0,05), effectue un test post-hoc (ex. test de Tukey HSD) pour identifier précisément quelles paires de groupes diffèrent significativement.

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(endog=df["math score"], groups=df["lunch"], alpha=0.05)
print(tukey)
```
