## III. Réduction de dimension par PCA

- Charge le second jeu de données depuis ex7data1.mat.
- Implémente la PCA pour trouver les composantes principales du jeu de données.
- Projette les données sur la première composante principale.
- Reconstruis les données à partir de la dimension réduite, et visualise le résultat.

```python
from scipy.io import loadmat
import numpy as np
import matplotlib.pyplot as plt

data = loadmat("ex7data1.mat")
X = data["X"]

# Normalisation des features
X_norm = (X - X.mean(axis=0)) / X.std(axis=0)

# Matrice de covariance puis SVD
cov_matrix = (X_norm.T @ X_norm) / X_norm.shape[0]
U, S, V = np.linalg.svd(cov_matrix)

# Projection sur la première composante principale
U_reduced = U[:, :1]
Z = X_norm @ U_reduced

# Reconstruction approximative
X_recovered = Z @ U_reduced.T

plt.scatter(X_norm[:, 0], X_norm[:, 1], label="Données normalisées")
plt.scatter(X_recovered[:, 0], X_recovered[:, 1], label="Données reconstruites (1 composante)")
plt.legend()
plt.title("PCA — projection et reconstruction")
plt.show()
```

> ⚠️ **Attention** : N'oublie pas de normaliser les features avant de calculer la matrice de covariance : la PCA est sensible à l'échelle des variables.
