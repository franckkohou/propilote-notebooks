## Les étapes d'un projet ML

- 1. Comprendre le problème : que prédit-on ? Classification (une catégorie) ou régression (un nombre) ?
- 2. Explorer (EDA) : distributions, valeurs manquantes, corrélations, déséquilibre des classes.
- 3. Prétraiter : imputer ou supprimer les valeurs manquantes, encoder les variables catégorielles, mettre à l'échelle les variables numériques.
- 4. Séparer entraînement et test (souvent 80/20) avant tout apprentissage, pour éviter les fuites de données.
- 5. Entraîner un ou plusieurs modèles.
- 6. Évaluer sur le jeu de test avec des métriques adaptées, puis interpréter.

> 💡 **Bonne pratique** : Un Pipeline applique exactement le même prétraitement à l'entraînement et au test, et évite d'« apprendre » la mise à l'échelle sur les données de test.

## Choisir ses métriques

| Problème | Métrique | Ce qu'elle mesure |
| --- | --- | --- |
| Classification | Accuracy | Part de prédictions correctes |
| Classification | Précision | Parmi les positifs prédits, combien le sont vraiment |
| Classification | Rappel | Parmi les vrais positifs, combien sont détectés |
| Classification | F1 | Moyenne harmonique de la précision et du rappel |
| Classification | Matrice de confusion | Le détail des bonnes et mauvaises prédictions |
| Régression | MAE | Erreur absolue moyenne, dans l'unité de la cible |

```python
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

> 💡 **À méditer** : En santé, un faux négatif (malade non détecté) coûte souvent plus cher qu'un faux positif : le rappel est alors une métrique clé, pas seulement l'accuracy.
